```python
import os
os.environ["PATH"] += ":/opt/homebrew/bin"

import graphviz
from graphviz import Digraph

dot = Digraph(format='png')
dot.attr(rankdir='TB', size='8,10')

# Styles
process_style = {'shape': 'box', 'style': 'rounded,filled', 'fillcolor': '#D6EAF8'}
decision_style = {'shape': 'diamond', 'style': 'filled', 'fillcolor': '#FAD7A0'}
mallow_style = {'shape': 'box', 'style': 'rounded,filled', 'fillcolor': '#D5F5E3'}
fixed_style = {'shape': 'box', 'style': 'rounded,filled', 'fillcolor': '#E8DAEF'}
final_style = {'shape': 'box', 'style': 'rounded,filled', 'fillcolor': '#A9DFBF'}

# Nodes
dot.node('start', 'Start', shape='oval', style='filled', fillcolor='#D5D8DC')

dot.node('inputs',
         'Inputs\n ts (DataFrame: date, mean_value)\n use_mallows (bool) · α = 3.0',
         **process_style)

dot.node('preprocess',
         'Temporal Pre-processing\nParse dates → t_numeric',
         **process_style)

dot.node('decision',
         'use_mallows?\n(knot selection strategy)',
         **decision_style)

dot.node('mallows',
         'Stepwise Knot Selection\nMallows CI (α=3.0)',
         **mallow_style)

dot.node('mallows_plot',
         'Plot Mallows CI Criterion\nhistory → n_knots opt.',
         **mallow_style)

dot.node('fixed',
         'Fixed Knot Count\nn_knots = n samples',
         **fixed_style)

dot.node('bspline',
         'B-Spline Basis Construction\nB = build_bspline_basis(...)',
         **process_style)

dot.node('lambda',
         'Regularisation Parameter Optimisation\nλ_opt = optimize_lambda_gcv(...)',
         **process_style)

dot.node('fit',
         'Penalised Spline Fitting\nŶ = penalized_spline_fit(...)',
         **process_style)

dot.node('final',
         'Final Fit Visualisation\nOriginal signal vs spline Ŷ',
         **final_style)

dot.node('return',
         'Return (dict)\ndates · values · t_numeric · n_knots · λ_opt · Ŷ · reg_std · history',
         shape='oval', style='filled', fillcolor='#D5D8DC')

# Edges
dot.edge('start', 'inputs')
dot.edge('inputs', 'preprocess')
dot.edge('preprocess', 'decision')

dot.edge('decision', 'mallows', label='True')
dot.edge('decision', 'fixed', label='False')

dot.edge('mallows', 'mallows_plot')
dot.edge('mallows_plot', 'bspline')

dot.edge('fixed', 'bspline')

dot.edge('bspline', 'lambda')
dot.edge('lambda', 'fit')
dot.edge('fit', 'final')
dot.edge('final', 'return')

#dot.render('run_regional_spline_flowchart', view=True)

# 1. Définissez le chemin du dossier où vous voulez enregistrer le graphique
# Exemple pour un dossier spécifique :
output_path = "rapport" 

# 2. On s'assure que le dossier existe, sinon Python le crée pour éviter une erreur
if not os.path.exists(output_path):
    os.makedirs(output_path)

# 3. On génère le rendu
# filename : le nom du fichier (sans extension)
# directory : le dossier de destination
# cleanup=True : supprime le fichier source .dot pour ne garder que le PNG
dot.render(filename='workflow_regional_spline', directory=output_path, cleanup=True, view=True)

print(f"Le graphique a été enregistré dans : {os.path.abspath(output_path)}")
```

# Diagramme Regression Locale

```python

# Configuration du graphe
dot = Digraph(format='png')
dot.attr(rankdir='TB', size='8,12')
dot.attr(label=r'\nAlgorithm: run_local_spline', labelloc='t', fontsize='16')

# --- STYLES ---
process_style = {'shape': 'box', 'style': 'rounded,filled', 'fillcolor': '#D6EAF8'}
decision_style = {'shape': 'diamond', 'style': 'filled', 'fillcolor': '#FAD7A0'}
parallel_style = {'shape': 'box', 'style': 'rounded,filled', 'fillcolor': '#EBF5FB'}
final_style = {'shape': 'box', 'style': 'rounded,filled', 'fillcolor': '#D5F5E3'}
io_style = {'shape': 'oval', 'style': 'filled', 'fillcolor': '#D5D8DC'}

# --- NOEUDS ---

# Entrée
dot.node('start', 'Start', **io_style)
dot.node('inputs', 
         'Inputs\nts (DF), lam_regional, std_regional, condition_cgv', 
         **process_style)

# Pré-traitement
dot.node('preprocess', 
         'Preprocessing\nParse dates → t_numeric\nn_knots = len(ts)\nstd_loc = std(values)', 
         **process_style)

dot.node('basis', 
         'B-Spline Basis Construction\nB = build_bspline_basis(t_numeric, n_knots)', 
         **process_style)

# Calculs Parallèles
dot.node('path_gcv', 
         'Spline GCV Path\n1. optimize_lambda_gcv\n2. penalized_spline_fit (Y_hat_gcv)', 
         **parallel_style)

dot.node('path_reg', 
         'Spline Regional Path\n1. Calculate lam_reg\n2. penalized_spline_fit (Y_hat_reg)', 
         **parallel_style)

# Décision
dot.node('decision', 'condition_cgv?', **decision_style)

# Résultat
dot.node('merge', 
         'Select Optimal Fit\nAssign lam_opt & Y_hat', 
         **final_style)

dot.node('viz', 
         'Final Visualisation\nPlot Original vs GCV vs Regional', 
         **final_style)

dot.node('return', 
         'Return (dict)\ndates, values, t_numeric, n_knots, lam, std_err, Y_hat', 
         **io_style)

# --- LIAISONS (EDGES) ---

dot.edge('start', 'inputs')
dot.edge('inputs', 'preprocess')
dot.edge('preprocess', 'basis')

# Division vers les deux calculs
dot.edge('basis', 'path_gcv')
dot.edge('basis', 'path_reg')

# Jonction vers la décision
dot.edge('path_gcv', 'decision')
dot.edge('path_reg', 'decision')

# Sorties de décision
dot.edge('decision', 'merge', label='True (GCV)')
dot.edge('decision', 'merge', label='False (Regional)')

dot.edge('merge', 'viz')
dot.edge('viz', 'return')

# --- SAUVEGARDE ---

output_path = "rapport"
if not os.path.exists(output_path):
    os.makedirs(output_path)

# Rendu
dot.render(filename='run_local_spline_flowchart', directory=output_path, cleanup=True, view=True)

print(f"Le diagramme local_spline a été enregistré dans : {os.path.abspath(output_path)}")
```
