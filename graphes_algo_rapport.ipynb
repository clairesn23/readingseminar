```python
import os
os.environ["PATH"] += ":/opt/homebrew/bin"

import graphviz
from graphviz import Digraph

dot = Digraph(format='png')
dot.attr(rankdir='TB', size='8,10')

# Styles
process_style = {'shape': 'box', 'style': 'rounded,filled', 'fillcolor': '#D6EAF8'}
decision_style = {'shape': 'diamond', 'style': 'filled', 'fillcolor': '#FAD7A0'}
mallow_style = {'shape': 'box', 'style': 'rounded,filled', 'fillcolor': '#D5F5E3'}
fixed_style = {'shape': 'box', 'style': 'rounded,filled', 'fillcolor': '#E8DAEF'}
final_style = {'shape': 'box', 'style': 'rounded,filled', 'fillcolor': '#A9DFBF'}

# Nodes
dot.node('start', 'Start', shape='oval', style='filled', fillcolor='#D5D8DC')

dot.node('inputs',
         'Inputs\n ts (DataFrame: date, mean_value)\n use_mallows (bool) · α = 3.0',
         **process_style)

dot.node('preprocess',
         'Temporal Pre-processing\nParse dates → t_numeric',
         **process_style)

dot.node('decision',
         'use_mallows?\n(knot selection strategy)',
         **decision_style)

dot.node('mallows',
         'Stepwise Knot Selection\nMallows CI (α=3.0)',
         **mallow_style)

dot.node('mallows_plot',
         'Plot Mallows CI Criterion\nhistory → n_knots opt.',
         **mallow_style)

dot.node('fixed',
         'Fixed Knot Count\nn_knots = n samples',
         **fixed_style)

dot.node('bspline',
         'B-Spline Basis Construction\nB = build_bspline_basis(...)',
         **process_style)

dot.node('lambda',
         'Regularisation Parameter Optimisation\nλ_opt = optimize_lambda_gcv(...)',
         **process_style)

dot.node('fit',
         'Penalised Spline Fitting\nŶ = penalized_spline_fit(...)',
         **process_style)

dot.node('final',
         'Final Fit Visualisation\nOriginal signal vs spline Ŷ',
         **final_style)

dot.node('return',
         'Return (dict)\ndates · values · t_numeric · n_knots · λ_opt · Ŷ · reg_std · history',
         shape='oval', style='filled', fillcolor='#D5D8DC')

# Edges
dot.edge('start', 'inputs')
dot.edge('inputs', 'preprocess')
dot.edge('preprocess', 'decision')

dot.edge('decision', 'mallows', label='True')
dot.edge('decision', 'fixed', label='False')

dot.edge('mallows', 'mallows_plot')
dot.edge('mallows_plot', 'bspline')

dot.edge('fixed', 'bspline')

dot.edge('bspline', 'lambda')
dot.edge('lambda', 'fit')
dot.edge('fit', 'final')
dot.edge('final', 'return')

dot.render('run_regional_spline_flowchart', view=True)
```
